```python
rs = bs.query_history_k_data_plus("sh.600000",
    "date,code,open,high,low,close,preclose,volume,amount,adjustflag,turn,tradestatus,pctChg,isST",
    start_date='2017-07-01', end_date='2017-12-31',
    frequency="d", adjustflag="3")
data_list = []
while (rs.error_code == '0') & rs.next():
    # 获取一条记录，将记录合并在一起
    data_list.append(rs.get_row_data())
data = pd.DataFrame(data_list, columns=rs.fields)
```

```python
data.head()
```

```python
data['date'] = pd.to_datetime(data['date'])
data.set_index('date', inplace = True) 
data.rename(columns={'close': 'price'}, inplace=True) 
data['price'] = list(map(float,data['price']))
data['volume'] = list(map(float,data['volume']))
```

```python
data['returns'] = np.log(data['price'] / data['price'].shift(1))
```

```python
data.dropna(inplace=True)
```

```python
total_dates = data.index.nunique()
split_index = int(total_dates * 0.7)
split_date = data.index.unique()[split_index]
train = data[:split_date]
test = data[split_date:]
```

```python
train['returns_dis'] = np.log(train['price'] / train['price'].shift(1))
train['returns_dis_cum'] = (train['returns_dis']).cumsum()+1
train.head()
```

```python
price_plot = ['returns_dis_cum']    
type(price_plot)
```

```python
for days in range(1, 23):
#     data['position_%d' % days] = np.sign(data['returns'].rolling(days).mean())
    price_plot.append('sty_cumr_%dd' % days)
    train['position_%dd' % days] = np.where(train['returns'].rolling(days).mean()>0,1,-1)
    train['strategy_%dd' % days] = train['position_%dd' % days].shift(1) * train['returns']
    train['sty_cumr_%dd' % days] = train['strategy_%dd' % days].cumsum()+1
```

```python
price_plot 
```

```python
train[price_plot].dropna().plot(
    title=' Multi Parameters Momuntum Strategy',
    figsize=(20, 12))
```

```python
mean_returns = []
std_returns = []
ratio_mean_std = []

for days in range(1,23):
    mean_return = train['strategy_%dd' % days].mean()
    std_return = train['strategy_%dd' % days].std()
    mean_returns.append(mean_return)
    std_returns.append(std_return)
    ratio_mean_std.append(mean_return / std_return)

print("Mean Returns:", mean_returns)
```

```python
print("Standard Deviations:", std_returns)
```

```python
print("Mean/Standard Deviation Ratios:", ratio_mean_std)
```

```python
x_labels = ['strategy_%dd' % days for days in range(1,23)]

# Plotting
plt.figure(figsize=(10, 6))
plt.bar(x_labels, mean_returns)
plt.xticks(rotation=45, ha='right')
plt.xlabel('Strategy')
plt.ylabel('Mean Return')
plt.title('Mean Returns for 29 Strategies')
plt.show()
```
